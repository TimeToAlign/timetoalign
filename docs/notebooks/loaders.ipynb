```python

from pathlib import Path
from timetoalign import ContinuousLogicalTimeline, DiscreteLogicalTimeline
from timetoalign.loader.score.music21 import Music21Loader
from timetoalign.loader.score.partitura import PartituraLoader
from timetoalign.loader.score.tsv import TSVLoader

DATA_DIR = Path(".").resolve().parents[1] / "tests" / "data" / "midi" / "score"
assert DATA_DIR.is_dir(), f"Data dir not found: {DATA_DIR}"
MS3_DIR = DATA_DIR / "ms3"
score_midi_dir = DATA_DIR / "score"
chopin_tsv_notes = MS3_DIR / "chopin_op10_no3.notes.tsv"
chopin_xml = DATA_DIR / "chopin_op10_no3.musicxml"
```

```python
l1 = TSVLoader()
l1.load(chopin_tsv_notes)
tsv_bundle = l1.bundle

l2 = PartituraLoader()
l2.load(chopin_xml)
pt_bundle = l2.bundle

l3 = Music21Loader()
l3.load(chopin_xml)
m21_bundle = l3.bundle

tsv_df = tsv_bundle.notes.to_dataframe()
pt_df = pt_bundle.notes.to_dataframe()
m21_df = m21_bundle.notes.to_dataframe()

tsv_notes = len(tsv_df[tsv_df["event_type"] == "Note"])
pt_notes = len(pt_df[pt_df["event_type"] == "Note"])
m21_notes = len(m21_df[m21_df["event_type"] == "Note"])

assert tsv_notes == 498
assert pt_notes == 498
assert m21_notes == 498
```
